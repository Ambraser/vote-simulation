## Data import and configuration

```python
# Folder containing election CSVs
DATA_FOLDER = "../cses_data"

# Candidate-count filter
# Keep only CSVs whose number of candidates is in this list.
# Set to None (or an empty list) to process every file in the folder.
CANDIDATE_FORKS: list[int] | None = [5, 6, 7, 8, 9]

# Voting rules
RULE_CODES: list[str] = [
    "AP_T05",
    "AP_T07",
    "BALD",
    "BLAC",
    "BORD",
    "BUCK_I",
    "BUCK_R",
    "CAIR",
    "COOM",
    "COPE",
    "HARE",
    "KIMR",
    "L4DV",
    "L6DV",
    "MJ",
    "MMIN",
    "NANS",
    "PLU1",
    "PLU2",
    "RV",
    "SCHU",
    "STAR",
    #    "VETO",
    "WOOD",
    "YOUN",
    "AP_K2",
    "AP_K3",
    "AP_KRP",
    "DODG_C",
]


OUTPUT_BASE = "../results"

COMPUTE_METRICS: bool = True

# Global reproducibility seed. Set to None to disable deterministic mode.
# The seeding is used to make the tie break when ordering candidates deterministic.
# Indeed, when candidates are graded the same by two voters, the profile generates randomly an order that tie breaks which in marginal cases changes the results.
REPRO_SEED: int | None = 161

summary, data = load_data_clean(
    data_folder=DATA_FOLDER,
    candidate_forks=CANDIDATE_FORKS,
)
```

## Simulation Run 

```python
total_result = run_simulations(
    selected=data, rule_codes=RULE_CODES, compute_metrics=COMPUTE_METRICS, repro_seed=REPRO_SEED, n_iterations=100
)
```

## Results analysis
Dissimilarity matrix

```python
total_result.plot_mean_distance_matrix(show=False, save_path=OUTPUT_BASE + "/mean_distance_matrix.png")
total_result.export_mean_distance_matrix_csv(save_path=OUTPUT_BASE + "/mean_distance_matrix.csv")
```

Metrics 

```python
total_result.plot_metrics_rules_matrix(show=False, save_path=OUTPUT_BASE + "/metrics_rules_matrix.png")
total_result.export_metrics_rules_matrix_csv(save_path=OUTPUT_BASE + "/metrics_rules_matrix.csv")
```

## Frequency of different outcomes

```python
total_filtered, df = analyze_results(total_result, OUTPUT_BASE, excluded_rules=None, candidate_forks=CANDIDATE_FORKS)
```

## Focus on Condorcet consistent rules

```python
CONDORCET_CONSISTENT_RULES = [
    "BALD",
    "BLAC",
    "COPE",
    "SCHU",
    "MMIN",
    "NANS",
    "YOUN",
    "DODG_C",
    "CAIR",
    "WOOD",
    "HARE",
]

# Keep only rules present in the current filtered set
cc_rules = [r for r in CONDORCET_CONSISTENT_RULES if r in total_result.rules]

print(f"Condorcet-consistent rules kept : {cc_rules}")

total_cc = total_result.filter_rules(cc_rules)

total_cc.plot_mean_distance_matrix(save_path=str(OUTPUT_BASE / "cc_mean_distance.png"))
```
